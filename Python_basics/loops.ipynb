```python
##while loop
"""the while loop continues to exceute a slong as the condition is true"""
c=0
while c<5:## it will print till 4 since it is not <= but it is <
    print(c)
    c+=1
```

```python
c=0;
while c%2==0:##even number condition
    print(c)
    c+=1
```

```python
#loop control statements
## break
"""the break statement exits the loop pematurely
"""
for i in range(10):
    if i==5:#when the condtioon is true it enter sthe break statemnt and it comes out of the loop
        break
    print(i)
#continue
#the continue statemnt skips the surrent iteration and it goes witht the next iteration
for j in range(10):
    if j%2==0:
        continue 
    print(j)
```

```python
"""pass statement is a null operation ,it does nothing"""
for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
#nested loops
#a loop inside a loop
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```
