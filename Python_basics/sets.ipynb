```python
#mathematical operations on sets
s={1,2,3,4}
s2={4,5,6,7,8,9}
u=s.union(s2)
print(u)
i=s.intersection(s2)
print(i)

s.intersection_update(s2)
print(s)
d=s.difference(s2)
print(d)
d1=s.difference_update(s2)
print(s)
```

```python
#symmetric difference
s={1,2,3,4,5}
s2={4,45,6,7,8,9}
sd=s.symmetric_difference(s2)
print(sd)
sd1=s.symmetric_difference_update(s2)
print(sd1)
```

```python
s1={1,2,3,4,5}
s2={3,4,5}
print(s1.issubset(s2))
print(s1.issuperset(s2))
```
