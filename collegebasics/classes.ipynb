```python
# create a class rectangle length and breadth to create methods to calcukate area and perimeter
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    def area(self):
        return self.length*self.breadth
    def p(self):
        return 2*(self.length+self.breadth)
n=int(input(""))
b=int(input(""))
a=int(input(""))
c=int(input(""))
rectangle1=rectangle(n,b)
rectangle2=rectangle(a,c)

print("Rectangle 1 area:",rectangle1.area())
print("Rectangle 2 area:",rectangle2.area())
print("Rectangle 1 perimeter:",rectangle1.p())
print("Rectangle 2 perimeter:",rectangle2.p())
```

```python
class BankAccount:
    def __init__(self, account_number, holder_name, balance=0):
        self.account_number = account_number
        self.holder_name = holder_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
        print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
            print("Balance:", self.balance)
        else:
            print("Insufficient balance")

    def display(self):
        print("Account number:", self.account_number)
        print("Holder name:", self.holder_name)
        print("Balance:", self.balance)


account = BankAccount(101, "Jeru", 1000)

account.display()
account.deposit(500)
account.withdraw(300)
account.withdraw(1500)
```

```python
class BankAccount:
    def __init__(self, account, balance):
        self.account = account
        self.__balance = balance

    def get(self):
        return self.__balance

    def set(self, amount):
        if amount >= 0:
            self.__balance = amount
            print(f"Balance set to: {self.__balance}")
        else:
            print("Amount cannot be negative")

    def deposit(self, amount):
        if amount < 0:
            print("Amount cannot be negative")
        else:
            self.__balance += amount
            print("The amount is:", self.__balance)

    def withdrawal(self, amount):
        if self.__balance < amount:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("The remaining amount is:", self.__balance)

    def display(self):
        print(f"Account no = {self.account}, Balance = {self.__balance}")


s1 = BankAccount("sb101", 500)

n = int(input("Enter the number of operations to be done: "))
for i in range(n):
    op = input("Enter which operation to perform: get, set, deposit, withdrawal, display: ")
    if op == "get":
        print(s1.get())
    elif op == "set":
        amt = int(input("Enter the new balance: "))
        s1.set(amt)
    elif op == "deposit":
        amt = int(input("Enter the deposit amount: "))
        s1.deposit(amt)
    elif op == "withdrawal":
        amt = int(input("Enter the withdrawal amount: "))
        s1.withdrawal(amt)
    elif op == "display":
        s1.display()
    else:
        print("Invalid operation")
```

```python
# the code is 
class person:
    def __init__(self,name):
        self.name=name
        print("Person constructor")
    def display(self):
        print("p1")

class student(person):
    def __init__(self, name,roll_no):
        super().__init__(name)
        self.roll_no=roll_no
        print("Student constructor")
    def display(self):
        super().display()  
        print("student")  
Student=student("Rahul",202)
Student.display()
print("n",Student.name)
print("r:",Student.roll_no)

```
