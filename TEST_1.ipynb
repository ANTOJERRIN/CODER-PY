```python
#bitwise
def bitwise():
    a=int(input("enter the value of a"))
    b=int(input("enter the value of b"))
    print("Bitwise And:",a&b,"\n")
    print("Bitwise Or:",a|b,"\n")
    print("Bitwise Xor:",a^b,"\n")
    print("Bitwise leftshift:",a<<b,"\n")
    print("Bitwise Rightshift:",a>>b,"\n")
bitwise()   
    
```
