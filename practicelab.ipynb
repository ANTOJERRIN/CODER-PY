```python
#this is all abt prog of using numpy and doing operations here 
import numpy as np #ivking the lib
#arr creation
a=np.array([10,20,30,40,50])
b=np.array([12,31,45,67,70])
#retruning the type of the array
print(type(a))
print(type(b))
#operations
print("addi",a+b)
print("mu",a*b)
#slicing
print("",a[:3])
print("",b[-2:])
#arrang
c=np.arange(1,10)
print("",c)
print("",c.reshape(3,3))
d=np.array([101,20,30,40,50])
print("",np.sum(d))
print("",np.prod(d))
print("",np.std(d))
print("",np.mean(d))
print("",np.median(d))
print("",np.min(d))
print("",np.max(d))
```

```python
#plotting grpahs
import matplotlib.pyplot as plt
import seaborn as sns
x=np.linspace(0,10,50)
y=np.sin(x)
plt.plot(x,y,label="sin(x)",color="red",marker="o")
plt.title("Line-plot sin function")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend()
plt.show()
```

```python
e=sns.load_dataset("iris")
sns.scatterplot(x="sepal_length",y="sepal_width",hue="species",data=e)
plt.title("sactterplot")
plt.show()
sns.boxplot(x="species",y="petal_length",data=e,palette="set2")
plt.title("BOX PLOT ")
plt.show()
```

```python
sns.boxplot(x="species",y="petal_length",data=e,hue="species",palette="Set2")
plt.title("BOX PLOT ")
plt.show()
```

```python
sns.pairplot(e,hue="species")
plt.suptitle("Pair plot -iris datatset",y=1.02)
plt.show()
```

```python
#iris dattset prohram
x=sns.load_dataset("iris")
print("\n",x.head())
print("dtat set info:")
print(x.info())
print("\n sum stats")
print(x.describe())
print("\nmissing val")
print(x.isnull().sum())
sns.histplot(x["sepal_length"],kde=True,bins=20,color="blue")
plt.title("distribtuion of sepal length")
plt.show()
sns.scatterplot(x="sepal_length",y="petal_length",hue="species",data=x)
plt.title("Sepal vs Petal Length by speices")
plt.show()
sns.pairplot(x,hue="species")
plt.suptitle("pair plot of iris features",y=1.02)
plt.show()
z=sns.load_dataset("iris")
num=z.select_dtypes(include='number')
cor=num.corr()
sns.heatmap(cor,annot=True,cmap="coolwarm")
plt.title("Correlation heatmap")
plt.show()
```
